```python
# EXIGÊNCIA DE CÓDIGO 1 de 8 — Mensagem de boas-vindas com nome
print("Bem-vindo ao sistema de gerenciamento de livros! Desenvolvido por Vitor Eduardo Blocki")

# EXIGÊNCIA DE CÓDIGO 2 de 8 — Lista vazia e variável de ID global
lista_livro = []  # Lista de dicionários
id_global = 0     # Será incrementado a cada novo livro

# EXIGÊNCIA DE CÓDIGO 3 de 8 — Função para cadastrar livro
def cadastrar_livro(id):
    print(">>> Cadastrar Livro <<<")
    nome = input("Digite o nome do livro: ")
    autor = input("Digite o nome do autor: ")
    editora = input("Digite o nome da editora: ")

    livro = {
        'id': id,
        'nome': nome,
        'autor': autor,
        'editora': editora
    }

    lista_livro.append(livro)
    print(f"Livro cadastrado com sucesso! ID: {id}")

# EXIGÊNCIA DE CÓDIGO 4 de 8 — Função para consultar livros
def consultar_livro():
    while True:
        print(">>> Consultar Livro <<<")
        print("1. Consultar Todos")
        print("2. Consultar por Id")
        print("3. Consultar por Autor")
        print("4. Retornar ao menu")

        opcao = input("Escolha a opção desejada: ")

        if opcao == '1':
            if not lista_livro:
                print("Nenhum livro cadastrado ainda.")
            else:
                print(">>> Todos os Livros Cadastrados <<<")
                for livro in lista_livro:
                    print(f"ID: {livro['id']} | Nome: {livro['nome']} | Autor: {livro['autor']} | Editora: {livro['editora']}")
        elif opcao == '2':
            try:
                id_busca = int(input("Digite o ID do livro: "))
                encontrado = False
                for livro in lista_livro:
                    if livro['id'] == id_busca:
                        print(f"ID: {livro['id']} | Nome: {livro['nome']} | Autor: {livro['autor']} | Editora: {livro['editora']}")
                        encontrado = True
                if not encontrado:
                    print("Livro com esse ID não encontrado.")
            except ValueError:
                print("Entrada inválida.")
        elif opcao == '3':
            autor_busca = input("Digite o nome do autor: ")
            encontrados = [livro for livro in lista_livro if livro['autor'].lower() == autor_busca.lower()]
            if encontrados:
                for livro in encontrados:
                    print(f"ID: {livro['id']} | Nome: {livro['nome']} | Autor: {livro['autor']} | Editora: {livro['editora']}")
            else:
                print("Nenhum livro encontrado para esse autor.")
        elif opcao == '4':
            break
        else:
            print("Opção inválida.")


# EXIGÊNCIA DE CÓDIGO 5 de 8 — Função para remover livro
def remover_livro():
    while True:
        try:
            id_remove = int(input("Digite o ID do livro que deseja remover: "))
            for livro in lista_livro:
                if livro['id'] == id_remove:
                    lista_livro.remove(livro)
                    print(f"Livro com id {id_global} removido com sucesso!")
                    return
            print("Id inválido.")
        except ValueError:
            print("Entrada inválida.")

# EXIGÊNCIA DE CÓDIGO 6 de 8 — Menu principal
while True:
    print("===== MENU PRINCIPAL =====")
    print("1. Cadastrar Livro")
    print("2. Consultar Livro")
    print("3. Remover Livro")
    print("4. Encerrar Programa")

    escolha = input("Escolha uma opção: ")

    if escolha == '1':
        id_global += 1
        cadastrar_livro(id_global)
    elif escolha == '2':
        consultar_livro()
    elif escolha == '3':
        remover_livro()
    elif escolha == '4':
        print("Programa encerrado. Até mais!")
        break
    else:
        print("Opção inválida. Tente novamente.")
```
