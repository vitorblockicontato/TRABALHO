```python
# EXIGÊNCIA DE CÓDIGO 1 de 7 — Mensagem de boas-vindas com nome
print("Bem-vindo ao sistema de cobrança da Copiadora! Desenvolvido por Vitor Eduardo Blocki.")

# Função para escolher o serviço (EXIGÊNCIA DE CÓDIGO 2 de 7)
def escolha_servico():
    while True:
        print("Escolha o serviço desejado:")
        print("DIG - Digitalização (R$1,10/página)")
        print("ICO - Impressão Colorida (R$1,00/página)")
        print("IPB - Impressão Preto e Branco (R$0,40/página)")
        print("FOT - Fotocópia (R$0,20/página)")
        servico = input("Digite o código do serviço (DIG/ICO/IPB/FOT): ")

        if servico == "DIG":
            return 1.10
        elif servico == "ICO":
            return 1.00
        elif servico == "IPB":
            return 0.40
        elif servico == "FOT":
            return 0.20
        else:
            print("Serviço inválido. Tente novamente.")  # EXIGÊNCIA DE SAÍDA DE CONSOLE 2 de 4

# Função para obter o número de páginas com desconto (EXIGÊNCIA DE CÓDIGO 3 de 7)
def num_pagina():
    while True:
        try:
            num = int(input("\nDigite o número de páginas: "))

            if num >= 20000:
                print("Quantidade de páginas não permitida.")  # EXIGÊNCIA DE SAÍDA DE CONSOLE 3 de 4
                continue
            elif num >= 2000:
                return num * 0.75  # 25% de desconto
            elif num >= 200:
                return num * 0.80  # 20% de desconto
            elif num >= 20:
                return num * 0.85  # 15% de desconto
            elif num > 0:
                return num  # sem desconto
            else:
                print("Número inválido. Tente novamente.")
        except ValueError:  # EXIGÊNCIA DE CÓDIGO 6 de 7
            print("Entrada inválida. Digite um número inteiro.")

# Função para escolher o serviço adicional (EXIGÊNCIA DE CÓDIGO 4 de 7)
def servico_extra():
    while True:
        print("Deseja algum serviço adicional?")
        print("1 - Encadernação Simples (R$15)")
        print("2 - Encadernação Capa Dura (R$40)")
        print("0 - Nenhum adicional (R$0)")
        adicional = input("Escolha a opção (1/2/0): ")

        if adicional == "1":
            return 15.00
        elif adicional == "2":
            return 40.00
        elif adicional == "0":
            return 0.00
        else:
            print("Opção inválida. Tente novamente.")

# ---------- CÓDIGO PRINCIPAL (EXIGÊNCIA DE CÓDIGO 5 de 7) ----------

# Chama as funções para obter os valores
valor_servico = escolha_servico()
paginas_com_desconto = num_pagina()
valor_extra = servico_extra()

# Calcula o total
total = (valor_servico * paginas_com_desconto) + valor_extra

# Exibe o valor total
print(f"\nValor total a pagar: R$ {total}")
print("Obrigado por usar o sistema!")
```
