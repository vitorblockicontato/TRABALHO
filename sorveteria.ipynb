```python
# EXIGÊNCIA DE CÓDIGO 1 de 8 – Mensagem de boas-vindas com nome e sobrenome
print("Bem-vindo à sorveteria Vitor Eduardo Blocki Ice!")

# Inicializa o acumulador de valor total do pedido
total_pedido = 0.0  # EXIGÊNCIA DE CÓDIGO 5 de 8

# Início do loop principal – EXIGÊNCIA DE CÓDIGO 6 e 7
while True:
    # Solicita o sabor ao cliente – EXIGÊNCIA DE CÓDIGO 2
    sabor = input("Digite o sabor desejado (CP para Cupuaçu / AC para Açaí): ")

    if sabor not in ['CP', 'AC']:
        print("Sabor inválido. Tente novamente.")  # EXIGÊNCIA DE SAÍDA DE CONSOLE 2
        continue  # EXIGÊNCIA DE CÓDIGO 7 - uso de continue

    # Solicita o tamanho ao cliente – EXIGÊNCIA DE CÓDIGO 3
    tamanho = input("Digite o tamanho desejado (P / M / G): ")

    if tamanho not in ['P', 'M', 'G']:
        print("Tamanho inválido. Tente novamente.")  # EXIGÊNCIA DE SAÍDA DE CONSOLE 3
        continue  # EXIGÊNCIA DE CÓDIGO 7 - uso de continue

    # EXIGÊNCIA DE CÓDIGO 4 – if/elif/else aninhado para todas combinações
    if sabor == 'CP':
        if tamanho == 'P':
            preco = 9.0
        elif tamanho == 'M':
            preco = 14.0
        else:  # G
            preco = 18.0
    elif sabor == 'AC':
        if tamanho == 'P':
            preco = 11.0
        elif tamanho == 'M':
            preco = 16.0
        else:  # G
            preco = 20.0

    # Adiciona o preço ao total
    total_pedido += preco  # EXIGÊNCIA DE CÓDIGO 5 de 8

    print(f"Pedido registrado: {sabor} tamanho {tamanho} - R$ {preco}")

    # Pergunta se o cliente quer mais alguma coisa – EXIGÊNCIA DE CÓDIGO 6
    resposta = input("Deseja pedir mais alguma coisa? (S/N): ")

    if resposta == 'N':
        break  # EXIGÊNCIA DE CÓDIGO 7 - uso de break

# Exibe o valor total do pedido
print(f"Valor total a pagar: R$ {total_pedido}")
print("Obrigado pela preferência!")
```
